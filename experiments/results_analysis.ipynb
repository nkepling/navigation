```python
import pandas as pd
import numpy as np
import matplotlib

%matplotlib inline

import sys
import os 
import pathlib
##### set up root dir
# Get the current working directory
file_path = os.path.join(os.getcwd(), 'results_analysis.ipynb')
# rootdir = os.path.dirname(file_path)

directory = os.path.dirname(file_path)

directory = pathlib.Path(directory)


```

```python
##### Summary stat functions and plotting########

def get_summary_stats(df:pd.DataFrame):
    summary_stats = df[["reward",
                        "steps",
                        "time",
                        "collisions",
                        "found_all_rewards",
                        "num_reward_blocks",
                        "num_obstacle_blocks"]].describe()

    return summary_stats


def get_mean_and_std_err(df:pd.DataFrame,varname:str):
    m = df[varname].mean()
    sderr = df[varname].std()/np.sqrt(df[varname].count())

    return m,sderr

    

```

```python
######### Baseline MCTS #############
basline_mcts_5x5 = pd.read_csv(directory / "results/baseline_mcts_5x5.csv")
print("Mean {0}, StdErr {1}".format(*get_mean_and_std_err(basline_mcts_5x5,"reward")))
get_summary_stats(basline_mcts_5x5)
```
